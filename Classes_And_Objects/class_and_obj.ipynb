## Private Attributes


```python
class BankAccount:
    def __init__(self,account_no,balance=0):
        self.__account_no=account_no
        self.__balance=balance

    def deposit(self,amount):
        self.__balance+=amount
        print(f'Amount {amount} deposited successfully')

    def withdraw(self,amount):
        if amount >=self.__balance:
            return 'Insufficient funds'
        self.__balance-=amount
        print(f'Amount {amount} withdred successfully')

    def check_Balance(self):
        return self.__balance

account=BankAccount('123456',5000)
print(account.check_Balance())
```

## Class Inheritance


```python
class Person:
    def __init__(self,first_name,last_name,age):
        self.first_name = first_name
        self.last_name = last_name
        self.age = age

    def get_person(self):
        print(f'{self.first_name} {self.last_name} {self.age}')

class Employee(Person):
    def __init__(self,first_name,last_name,age,employee_id):
        super().__init__(first_name,last_name,age)
        self.employee_id = employee_id

    def get_id(self):
        return f'{self.employee_id}'

emp=Employee('Divyansh','Singh',22,'123456')
print(emp.get_id())
```

## Method Overriding

```python
class Employee(Person):
    def __init__(self,first_name,last_name,age,employee_id):
        super().__init__(first_name,last_name,age)
        self.employee_id = employee_id

    def get_id(self):
        return f'{self.employee_id}'

    def get_person(self):
        return f'{self.employee_id} {self.first_name} {self.last_name} {self.age}'

emp=Employee('Divyansh','Singh',22,'123456')
print(emp.get_id())
print(emp.get_person())
```

## Class Composition

```python
class Address:
    def __init__(self,street,city,postal_code):
        self.street = street
        self.city = city
        self.postal_code = postal_code

    def __str__(self):
        return f'{self.street} {self.city} {self.postal_code}'

class Person:
    def __init__(self,first_name,last_name,address):
        self.first_name = first_name
        self.last_name = last_name
        self.address = address

    def __str__(self):
        return f'{self.first_name} {self.last_name} {self.address}'

address=Address('Rjmp','Lucknow',226017)
person=Person('Divyansh','Singh',address)

print(person.__str__())
```

## Class With Properties

```python
class Rectangle:
    def __init__(self,width,height):
        self.__width = width
        self.__height = height

    @property
    def width(self):
        return self.__width

    @width.setter
    def width(self,new_width):
        self.__width = new_width

    @property
    def height(self):
        return self.__height

    @height.setter
    def height(self,new_height):
        self.__height = new_height
```

## Abstract Methods

```python
from abc import ABC,abstractmethod
import math

class Shape(ABC):

    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius

    def area(self):
        return math.pi*self.radius**2
```

## Operator Overloading

```python
class Vector:
    def __init__(self,x,y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x+other.x,self.y+other.y)

    def __sub__(self, other):
        return Vector(self.x-other.x,self.y-other.y)

    def __mul__(self, other):
        return Vector(self.x*other.x,self.y*other.y)

v1=Vector(1,2)
v2=Vector(3,4)
print(v1+v2)
print(v1*v2)
```
