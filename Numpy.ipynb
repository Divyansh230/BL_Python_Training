````python
import numpy as np

# create random array
arr = np.random.random(10)

print("Original array elements:")
print(arr)

# set precision to 3
np.set_printoptions(precision=3)

print("\nPrint array values with precision 3:")
print(arr)

# convert to list
lst = arr.tolist()

print("\nConverted Python list:")
print(lst)
````

````python



arr = np.array([1.6e-10, 1.6, 1200, 0.235])

print("Original array elements:")
print(arr)

# suppress scientific notation + set precision
np.set_printoptions(suppress=True, precision=3)

print("\nPrint array values with precision 3:")
print(arr)
````
