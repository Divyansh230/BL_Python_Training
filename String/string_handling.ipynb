```python
## Count vowels
count=0
s='Avengers Doomsday'
for c in s:
    if c in 'aeiou':
        count+=1
print(count)
```

```python
## First non repeating Character
s='aaabccde'
freq={c:s.count(c) for c in s}

for k,v in freq.items():
    if v==1:
        print(k)
        break

```

```python
## Anagram check
a='listen'
b='istlen'
print(sorted(a)==sorted(b))
```

```python
## Removing Duplicates
s='aaabbbcccde'
ls=[]
for ch in s:
    if ch not in ls:
        ls.append(ch)
print("".join(ls))
```

```python
## Longest Substring Without Repeating Character
s = "abcabcbb"
map={}
start,end=0,0
max_len=0

while end<len(s):
    if s[end] in map and map[s[end]]>=start:
        start=map[s[end]]+1
    max_len=max(max_len,end-start+1)
    end+=1

print(max_len)
```

```python
## String Compression
s= "aaabbc"
map={x:s.count(x) for x in s}
ls=[]
for k,v in map.items():
    ls.append(k+str(v))
print("".join(ls))

```

```python
## Checking String Rotation of other or not
s1 = "abcd"
s2 = "cdab"

print(len(s1)==len(s2) and s2 in s1+s2)
```

```python
##Longest Palindromic Substring
s='babad'

def expan(l,r):
    while l>=0 and r<len(s) and s[l]==s[r]:
        l-=1
        r+=1
    return s[l+1:r]
ans=''
for i in range(len(s)):
    t1=expan(i,i)
    t2=expan(i,i+1)
    if len(t1)>len(ans):
        ans=t1
    if len(t2)>len(ans):
        ans=t2
print(ans)
```
