```python
import re
```

```python
s='hello123456'
print(re.findall(r'\d',s))

## \d-> it represent the digit
```

## Regex Methods
1. re.search(): first match
2. re.match(): check first string
3. re.findAll(): return the all match
4. re.sub():replace


```python
## Extract all digit
s="abc123xyz45"
print(re.findall(r'\d',s))
```

```python
## Count words
s='Hello World, I am learning Python'
print(len(re.findall(r'\w',s)))
```

```python
## Validating Mobile Number
pattern=r'^[6-9]\d{9}$'
print(bool(re.match(pattern,'8791314283')))
```

```python
## Validating User Name
pattern=r'^[a-zA-Z0-9_]{5,15}$'
print(bool(re.match(pattern,'divyansh_123')))
```

```python
## Password Validation
pattern =r'^(?=.*[a-z])(?=.*[A-Z])(?=.*[0-9]){8,}$'
print(bool(re.match(pattern,s)))
```

```python
## Email Validation
patter=r'^[a-z0-9]+@[a-z]+\.[a-z]+$'
print(bool(re.match(patter,'divyans125@gmail.com')))
```
