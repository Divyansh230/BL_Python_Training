# Data Cleaning


```python
df=pd.DataFrame(np.random.randint(1,100,size=(5,3)),columns=['A','B','C'])
df
```

```python
df.iloc[0,1]=np.nan
df.iloc[2,2]=np.nan

df
```

```python
df.fillna(df.mean(),inplace=True)
df
```

```python
df=pd.DataFrame(np.random.randint(1,100,size=(6,4)),columns=['A','B','C','D'])
df
```

```python
df.iloc[1, 2] = np.nan
df.iloc[3, 0] = np.nan
df.iloc[5, 1] = np.nan
```

```python
df
```

```python
df.dropna(axis=1,inplace=True)
df
```
