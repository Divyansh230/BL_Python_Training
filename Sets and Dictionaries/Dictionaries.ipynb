```python
## Counting frequency of Character in the String
def freq(s):
    frequency={}
    for char in s:
        if char not in frequency:
            frequency[char]=1
        else:
            frequency[char]+=1

    return frequency

s="Raju Ban Gaya Gentle Man"
print(freq(s))
```
