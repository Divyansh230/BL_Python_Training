```python
## Creating and Accessing the Sets

st=set(range(1,11))
print(st)
```

```python
st.add(11)
print(st)

st.remove(1)
print(st)
```

```python
set1={1,2,3,4,5}
set2={2,4,6,8,10}

##Union
print(f'Union:{set1.union(set2)}')
print(f'Intersection:{set1.intersection(set2)}')
print(f'Difference:{set1.difference(set2)}')
print(f'Symmetric_difference:{set1.symmetric_difference(set2)}')
```

```python
##Set Comprehension

set_sq={x**2 for x in range(10)}
print(set_sq)
```

```python
##Storing the Square of the Even numbers
set_sq_even={x for x in set_sq if x%2==0}
print(set_sq_even)
```

```python
## Superset and Subset
set_a={1,2,3,4,5}
set_b={2,3,4}

print(set_a.issubset(set_b))
print(set_b.issuperset(set_a))
```

```python
s=set(range(10))
while len(s)!=0:
    s.pop()
    print(s)
```
